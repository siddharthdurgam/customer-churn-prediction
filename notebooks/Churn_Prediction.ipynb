GitHub-safe data preparation and baseline modelling notebook.

## 1. Imports

```python
import sys
from pathlib import Path

import pandas as pd
from sklearn.model_selection import train_test_split

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "src").exists():
    PROJECT_ROOT = Path.cwd().parent

sys.path.append(str(PROJECT_ROOT / "src"))

from data_preprocessing import load_csv, prepare_features_and_target
from evaluate import print_classification_metrics
```

## 2. Load data

Place an exported churn dataset at `data/raw/churn_data.csv`. Raw/source data is intentionally not committed to this repository.

```python
DATA_PATH = PROJECT_ROOT / "data" / "raw" / "churn_data.csv"

df = load_csv(DATA_PATH)
display(df.head())
print(f"Shape: {df.shape}")
```

## 3. Inspect the target distribution

```python
if "Customer_Status" in df.columns:
    display(df["Customer_Status"].value_counts())
else:
    print("Customer_Status column is not available in this dataset.")
```

## 4. Preprocess features

```python
X, y = prepare_features_and_target(df)

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
display(X.head())
```

## 5. Train / test split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y,
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

## 6. Optional: SMOTE balancing

Use SMOTE only on the training split to avoid data leakage into the test set.

```python
from balance_data import balance_dataset

X_train_balanced, y_train_balanced = balance_dataset(X_train, y_train)

print("Original training distribution:")
print(y_train.value_counts())

print("\nBalanced training distribution:")
print(pd.Series(y_train_balanced).value_counts())
```
